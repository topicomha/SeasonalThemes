# Add Simlink

### Linux
My Linux path to all oh-my-posh themes: `/home/david/.oh-my-posh-themes/` Example Theme File:  `davids-november.omp.json`

Path to repo: `/home/david/Code/seasonal-theming/[Month]/` Example Theme File: `davids-november.omp.json`

```powershell
$Month = "May"
$Path = "/home/david/.oh-my-posh-themes/davids-$Month.omp.json"
$Target = "/home/david/Code/seasonal-theming/$Month/davids-$Month.omp.json"

$Month
$Path
$Target
```

```powershell

New-Item -ItemType SymbolicLink -Path $Path -Target $Target
```

Powershell doesn't work on linux. So here is the linux script to create it. 

Also, I don't think powershell works with relative paths. Use absolute paths.

```powershell
sudo ln -s $Target $Path 
```
